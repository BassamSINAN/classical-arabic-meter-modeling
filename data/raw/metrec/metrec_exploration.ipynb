# MetRec: exploration of the raw files

MetRec (Al-shaibani, Alyafeai & Ahmad, 2020, *Data in Brief* 33) is a corpus of
Arabic verses labelled with their meter, collected by the authors from the
Aldiwan website.

This notebook explores the raw files as published, without modifying them.
The files are downloaded with `download_metrec.py`, in this folder.

````python
import pathlib

# The notebook sits in the same folder as the data files.
DATA = pathlib.Path(".")
for path in sorted(DATA.glob("*.txt")):
    print(path.name)
````

## `labels.txt`

The meter names, one per line.

````python
labels_path = DATA / "labels.txt"
labels = labels_path.read_text(encoding="utf-8").split()

print(f"Size: {labels_path.stat().st_size} bytes")
print(f"Lines: {len(labels)}")
print()
for number, name in enumerate(labels):
    print(f"{number:>2}  {name}")
````

- The file is tiny: one meter name per line, nothing else.
- According to the authors, the classes are numbered from 0 to 13.
- Two of the sixteen *buḥūr* are absent: the *muḍāriʿ* and the *muqtaḍab*.
  The authors explain that not all meters are commonly used, and that some
  Arab scholars consider these two non-existent.

## `train.txt` and `test.txt`

Plain UTF-8 text files, one verse per line, with no header. Each line has the
form:

```
<meter number> <ṣadr> # <ʿaǧuz>
```

The split between `train` and `test` was made by the authors.

````python
for name in ["train.txt", "test.txt"]:
    lines = (DATA / name).read_text(encoding="utf-8").splitlines()
    print(f"{name}: {len(lines)} lines")
````

| File | Verses |
|---|---|
| `train.txt` | 47,124 |
| `test.txt` | 8,316 |
| **Total** | **55,440** |

The test set is 15% of the total.

````python
with open(DATA / "train.txt", encoding="utf-8") as file:
    for _ in range(5):
        print(file.readline(), end="")
````

````python
with open(DATA / "test.txt", encoding="utf-8") as file:
    for _ in range(3):
        print(file.readline(), end="")
````

## Observations

- The meters are mixed, in an apparently random order.
- Vocalisation varies from one verse to the next: some verses are almost
  fully vocalised, others not at all.
- Some verses are *mudawwar*: a word straddles the *ṣadr* and the *ʿaǧuz*,
  and the `#` splits it in two. Example: وَالصَو # نِ, for وَالصَّوْنِ. The `#`
  therefore does not always fall between two words.

**Challenges related to vocalisation.**

   whether a verse is vocalised enough to be scanned, which calls for research
   and analysis, and then to count how many such verses the corpus contains.
   If that number turns out to be too small, we move on to the second challenge.
